```python
#W7A1
def binary_search(arr, left, right, target):
    if left > right:
        return -1 

    mid = (left + right)// 2

    if arr[mid] == target:
        return mid 
    elif arr[mid] > target:
        return binary_search(arr, left, mid - 1, target)
    else: 
        return binary_search(arr, mid + 1, right, target)

arr = list(map(int,input().split()))
target = int(input())
result = (binary_search(arr,0, len(arr) -1 , target))
print(result)
```

```python
#W7A2
def count_occurences(arr,x):
    count = 0
    for value in arr:
        if value == x:
            count += 1
    return count 
arr = list(map(int,input().split()))
x = int(input())
print(count_occurences(arr,x))
```

```python
#W7A3
def bubble_sort(arr):
    n = len(arr)
    for i in range (n-1):
        for j in range (n-1-i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr 
arr = list(map(int,input().split()))
print(bubble_sort(arr))
```

```python
#W7A3
def selection_sort(arr):
    n = len (arr)
    for i in range (n-1):
        min_index = i
        for j in range (i+1,n):
            if arr[j] < arr[min_index]:
                min_index = j
            arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
arr = list(map(int,input().split()))
print(selection_sort(arr))
```

```python
#W7A4
def most_frequent(arr):
    count = {}
    max_count = 0
    result = None
    for x in arr:
        count[x] = count.get(x,0) + 1
        if count[x] > max_count:
            max_count = count[x]
            result = x 
    return result 
arr = list(map(int,input().split()))
print(most_frequent(arr))
```

```python
#W7A5
def count_pairs(arr,X):
    count = 0
    n = len (arr)
    for i in range (n):
        for j in range (i+1,n):
            if arr[i] + arr[j] == X:
                count += 1
    return count 
arr = list(map(int,input().split()))
X = int(input())
print(count_pairs(arr,X))
```

```python
#W7A6
n = int(input('Nhập và số danh sách con:'))
nested_list = []
for i in range (n):
    sub_list = list(map(int,input().split()))
    nested_list.append(sub_list)
flat_list = []
for sub in nested_list:
    flat_list.extend(sub)
print(flat_list)
```

```python
#W7A7
nums = list(map(int,input().split()))
n = len(nums)
length = [1] * n
for i in range (n):
    for j in range (i):
        if nums[j] < nums[i]:
            if length[j] + 1 > length[i]:
                length[i] = length[j] + 1
print(max(length))
```

```python
#W7A16
def beautiful_array(arr):
    checked = []

    for x in arr:
        if x in checked:
            continue

        cnt = 0
        for y in arr:
           if y == x:
               cnt += 1

        if cnt != x:
            return "No"

    return "YES"

arr = list(map(int,input().split()))
print(beautiful_array(arr))
                
```

```python
#W11A1
def reverse(path: str) -> int: 
    with open(path, "r") as f:   
        n_str = f.read().strip() 
    return int(n_str[::-1])  

n = input("Nhập một số nguyên dương n: ")

with open("number.txt", "w") as f: 
    f.write(n)

print(reverse("number.txt"))
```

```python
#W11A2
def canWinNim(path: str) -> bool:
    with open(path, 'r') as f:
        n = int(f.read().strip())
    return n % 4 != 0
with open('bai2.txt','w') as f:
    f.write("7")
print(canWinNim('bai2.txt'))
```

```python
#W11A9
import csv
def grade10(path: str):
    with open(path) as f:
        reader = csv.DictReader(f)
        for row in reader:
            if row['THCS4'] == '10':
                print(row['Ho'] + " " + row['Ten'])
with open('bai9.txt','w', encoding = 'utf-8') as f:
    f.write("STT,Ho,Ten,Ngay sinh,Dai so,THCS4,THCS1\n1,Truong Vo,Ky,10/10/1999,8,10,9\n2,Chu Chi,Nhuoc,13/9/1999,10,10,10\n")


grade10('bai9.txt')
```

```python
#W11A5
def count(path: str, character: str) -> int:
    with open(path) as f:
        text = f.read()
    return text.count(character)

with open('bai5.txt','w', encoding = 'utf-8') as f:
    f.write("We accept the love we think we deserve\ne\n")

print(count('bai5.txt', 'e'))
```

```python
#W11A3
def housesOfHogwarts(path: str):
    with open(path, 'r', encoding='utf-8') as f:
        lines = f.read().strip().splitlines()
    
    n = int(lines[0])
    names = lines[1:1+n]
    
    for name in names:
        print(name)
with open('hogwarts.txt','w', encoding = 'utf-8') as f:
    f.write("5\nHarry Potter\nHermione Granger\nRon Weasley\nDraco Malfoy\nLuna Lovegood")
housesOfHogwarts("hogwarts.txt")
```

```python
#W11A4
def areAnagrams(path1: str, path2: str) -> bool:
    def clean(s):
        import re
        return ''.join(sorted(re.sub(r'[^a-z]', '', s.lower())))
    with open(path1) as f1, open(path2) as f2:
        s1, s2 = f1.read(), f2.read()
    return clean(s1) == clean(s2)
with open('file1.txt','w') as f:
    f.write("Tom Marvolo Riddle")
with open('file2.txt','w') as f:
    f.write(" I am Lord Voldermort!")
print(areAnagrams('file1.txt','file2.txt'))
```

```python
#W11A6
def countCharacter(path: str, letter: str) -> int:
    with open(path, 'r', encoding='utf-8') as f:
        text = f.read()
    
    text = text.lower()
    letter = letter.lower()
    
    return text.count(letter)
with open('bai6.txt','w') as f:
    f.write("Everyone derserves a second shot.")

print(countCharacter("bai6.txt","e"))
```

```python
#W11A8
def write_fibonacci_words(n: int):
    if n < 0 or n > 20:
        return
    
    fib = []
    fib.append("a")   # f(0)
    if n >= 1:
        fib.append("b")   # f(1)

    for i in range(2, n+1):
        fib.append(fib[i-1] + fib[i-2])

    with open("output.txt", "w", encoding="utf-8") as f:
        for i in range(n+1):
            f.write(f"f({i}) = {fib[i]}\n")


# Ví dụ chạy
n = int(input())
write_fibonacci_words(n)
```

```python
#W7A16
def min_remove_to_beautiful(arr):
    cnt = {}
    for x in arr:
        if x in cnt:
            cnt[x] += 1
        else:
            cnt[x] = 1

    remove = 0
    for x in cnt:
        if cnt[x] < x:
            remove += cnt[x]
        elif cnt[x] > x:
            remove += cnt[x] - x

    return remove

arr=list(map(int,input().split()))
print(min_remove_to_beautiful(arr))
```

```python
#W6A16
n, m = map(int, input().split())
lst = []
for i in range (n):
    lst.append(list(map(int, input().split())))
for i in range (n):
    for j in range (m):
        print(f"{lst[i][j]:>4}", end ="")
    print()
```

```python
#W6A18
n, m, k = list(map(int, input().split()))
lst = []
th = 0
tc = 0
for i in range (n):
    lst.append(list(map(int, input().split())))
for i in range (n):
        th += lst[k-1][i]
for i in range (m):
        tc += lst[k-1][i]
print(th, tc)
```

```python
#W6A19
lst = list(map(int, input().split()))
def pos_max(lst):
    pmax = lst[0]
    pos = 0
    for i in range(1, len(lst)):
        if lst[i] > pmax:
            pmax = lst[i]
            pos = i + 1
    return pos
print(pos_max(lst))
```

```python
#W6A20
lst = list(map(int, input().split()))
k = int(input())
pos = 0
for i in range( len(lst)):
    if lst[i] == k:
        pos = i +1
        break
    else: pos = -1
print(pos)
```

```python
#W6A21
lst = list(input().split())
d = {}
for i in range(len(lst)):
    word = lst[i]
    if word not in d:
        d[word] = []
    d[word].append(i + 1)
for j in d:
    d[j] = tuple(d[j]) 
print(d)
```

```python
#W6A22
lst = list(input().split())
k = max(lst)
for i in range (len(lst)):
    if k == lst[i]:
        print(i+1)
        break
```

```python
#W6A23
lst = list(map(int, input().split()))
k = int(input())
nlst = []
for i in range (len(lst)):
    if lst[i] >= k:
        nlst.append(lst[i])
print(nlst)
```

```python
#W6A24
lst = list(map(int, input().split()))
zr = []
other = []
for i in range (len(lst)):
    if lst[i] == 0:
        zr.append(lst[i])
    if lst[i] != 0:
        other.append(lst[i])
print( zr + other)
```

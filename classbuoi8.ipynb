```python
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return (self.length + self.width) * 2
l,w = map(int,input().split())
r = Rectangle (l,w)
print(r.area())
print(r.perimeter())
```

```python
class Student:
    def __init__(self,name,singing,dancing):
        self.name = name
        self.singing = singing
        self.dancing = dancing
    def average(self):
        return (self.singing + self.dancing) / 2
    def result(self):
        if self.average() >= 5.0:
            return("Pass")
        else:
            return("Fail")
name = input()
s,d = map(float,input().split())
s = Student(name,s,d)
print(f"{s.average():.2f}")
print(s.result())
```

```python
#W8A4
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def set_numbers(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def subtract(self):
        return self.a - self.b
    def product(self):
        return self.a * self.b
    def divide(self):
        if self.b == 0:
            return "Không thể chia cho 0"
        return self.a / self.b
    def power(self):
        return self.a ** self.b
    def mod(self):
        if self.b == 0:
            return "Không thể chia cho 0"
        return self.a % self.b
a,b = map(int,input().split())
calc = Calculator(a,b)
while True:
    print("MENU")
    print("1.ADD")
    print("2.SUBTRACT")
    print("3.PRODUCT")
    print("4.DIVIDE")
    print("5.POWER")
    print("6.MOD")
    print("7.SET NUMBERS")
    choice = input("Chọn phép toán (1-7):")
    if choice == "1":
        print(calc.add())
    elif choice == "2":
        print(calc.subtract())
    elif choice == "3":
        print(calc.product())
    elif choice == "4":
        print(calc.divide())
    elif choice == "5":
        print(calc.power())
    elif choice == "6":
        print(calc.mod())
    elif choice == "7":
        a,b = map(int,input().split())
        calc.set_number(a,b)
    else:
        print("Lựa chọn không hợp lệ!")
    stop = input("Bạn có muốn thoát?")
    if stop in ["yes","Yes","YES"]:
        break
```

```python
#W8A6
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def scale(self,k):
        self.width *= k 
        self.height *= k  
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return (self.width + self.height) * 2
w,h,k = map(int,input().split())
r = Rectangle(w,h)
r.scale(k)
print(r.area())
print(r.perimeter())
```
